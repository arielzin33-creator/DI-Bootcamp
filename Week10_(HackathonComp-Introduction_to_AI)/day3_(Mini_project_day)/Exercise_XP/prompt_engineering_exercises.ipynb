## 🌟 Exercise 1: Rewrite and Optimize a Vague Prompt

**Scenario:** Manager's vague prompt: *"Write something about productivity tips."* — meant to promote FlowNest, a focus app, on LinkedIn.

### Task 1 — Three critical issues with the original prompt

1. **No role/brand voice specified.** The model doesn't know it should write *as* a productivity/focus-app brand (FlowNest) with a specific tone (e.g., calm, encouraging, science-backed). Without this, the output could sound generic, overly corporate, or off-brand.
2. **No target audience defined.** "Productivity tips" for whom — students, remote workers, executives, ADHD users? Without an audience, the tips will be too broad to resonate or drive engagement/clicks toward FlowNest.
3. **No platform, format, or goal constraints.** LinkedIn posts have specific conventions (hook line, short paragraphs/bullets, a CTA, character limits, hashtags). The prompt doesn't mention FlowNest at all, doesn't ask for a call-to-action, and gives no length or format — the model could return a generic essay instead of a scroll-stopping post.

### Task 2 — Rewritten prompt (Role + Instruction + Format/Tone/Length)

> **Act as** a senior social media copywriter for FlowNest, a focus and deep-work app for busy professionals.
>
> **Give 3 actionable, non-obvious productivity tips** that tie naturally into how FlowNest helps people focus (e.g., timers, distraction blocking, deep-work sessions). Each tip should include one sentence of practical advice and, where relevant, a subtle mention of how FlowNest supports it — without sounding like a hard sales pitch.
>
> **Format constraints:** Write it as a LinkedIn post. Use a short attention-grabbing hook as the first line, then 3 bullet points (one per tip), and end with a soft call-to-action inviting readers to try FlowNest. Keep the total post under 200 words. Tone: encouraging, confident, and conversational — avoid corporate jargon and emojis except for at most 2.


### Task 3 — Reusable closing line

> *Use this format whenever we promote productivity tools or features on social media: role as FlowNest copywriter, 3 actionable tips tied to the product, LinkedIn post format with hook + bullets + soft CTA, under 200 words, encouraging conversational tone.*

## 🌟 Exercise 2: Multi-Part Prompt for Quiz Generation

**Scenario:** A 7th-grade science teacher has a short article on volcanic eruptions and wants ChatGPT to create quiz content for Google Slides.

### Task 1 — Complete prompt

> **Act as** a middle school science teacher's assistant creating engaging lesson content for 11–13-year-old students.
>
> I will give you a short article about volcanic eruptions. Using **only** the information in the article:
>
> 1. **Summarize the article in exactly 2 simple bullet points**, using vocabulary appropriate for 11–13-year-olds (short sentences, no jargon, explain any necessary science terms simply).
> 2. **Generate 3 age-appropriate multiple-choice questions** based on the article. Each question must have exactly 1 correct answer and 2 plausible but incorrect distractors (label them A, B, C). Mark the correct answer clearly at the end of each question.
>
> **Tone:** Friendly, encouraging, and simple — like a teacher talking to curious kids. Avoid complex or technical vocabulary; if a science term is required, briefly define it in plain language.
>
> **Output format:** Structure the output as separate slides for Google Slides — one slide title "Summary" with the 2 bullet points, followed by one slide per question (title "Question 1", "Question 2", "Question 3") each containing the question, the 3 answer choices, and the correct answer noted separately below a divider (so it can be hidden from students if needed).
>
> Here is the article: [PASTE ARTICLE TEXT HERE]

### Task 2 — Why this prevents irrelevant or low-quality output (vs. "Make a quiz for kids about this article.")

1. **Grounding in the source material.** By explicitly instructing the model to use *only* the article's content, we prevent it from inventing unrelated volcano facts (hallucination) or pulling in generic textbook trivia not covered in class.
2. **Explicit structure and counts.** Specifying "2 bullet points," "3 questions," "1 correct + 2 distractors" removes ambiguity — the vague prompt could return 1 question or 10, in paragraph form, with no clear answer key, making it unusable for a slide deck.
3. **Audience-calibrated tone and vocabulary.** Naming the exact age range (11–13) and requiring simple vocabulary stops the model from producing content pitched at adults or using overly technical volcanology terms that would confuse the target students — the vague prompt has no such constraint and risks output that's either too babyish or too advanced.

## 🌟 Exercise 3: Add Context, Get Better Results

**Scenario:** An intern's prompt ("Summarize this report.") produces a generic summary with no useful figures for a 3-minute monthly finance update.

### Task 1 — Missing context table

| Context Type | Is it Missing? | What Should Be Added? |
|---|---|---|
| **Role** | Yes | "Act as a financial analyst preparing an executive briefing." |
| **Audience** | Yes | "This is for a non-technical executive team with limited time." |
| **Purpose** | Yes | "The summary will be read aloud in a 3-minute monthly finance update presentation." |
| **Input Source** | Partially | Clarify exactly which report/section/time period is being summarized (e.g., "the attached Q2 monthly finance report, pages 1–5"). |
| **Format/Style** | Yes | "Present as 3 key takeaways, each with one supporting figure, in plain language — no jargon." |
| **Constraints** | Yes | "Limit to 3 bullet points, must be readable aloud in under 3 minutes (~350 words max), include at least one specific number per point." |

### Task 2 — Rewritten prompt (4+ context types included)

> **Act as** a financial analyst preparing an executive briefing (**Role**).
>
> This summary is for a **non-technical executive team** who will hear it read aloud during a **3-minute monthly finance update presentation** (**Audience** + **Purpose**).
>
> Using the attached **Q2 monthly finance report** as your only source, summarize it (**Input Source**).
>
> **Limit your summary to exactly 3 key takeaways**, each with one supporting data point (a specific number, percentage, or dollar figure from the report). Keep the total under 350 words so it can be read aloud in under 3 minutes. Use plain, non-technical language and avoid financial jargon (**Format/Style** + **Constraints**).

## 🌟 Exercise 4: Match Prompt to Purpose

**Scenario:** Building prompt templates for an e-commerce customer support chatbot.

**Style chosen: Functional** — customer support needs precise, reliable task execution (e.g., order status, return eligibility) rather than brainstorming or free-flowing chat.

### Prompt (Functional style)

> You are a customer support assistant for an e-commerce store. A customer will describe an issue (e.g., order status, return request, refund question). Your task:
>
> 1. Identify the customer's intent from their message (order status / return / refund / product question / other).
> 2. Extract any relevant details they provided (order number, product name, dates).
> 3. If information is missing that's required to resolve the request (e.g., no order number given), ask for exactly that missing piece — nothing else.
> 4. If all needed information is present, respond with a precise, step-by-step resolution using only the store's official policies provided below. Do not speculate or invent policy details.
> 5. Keep responses under 100 words, in a polite and professional tone, and end with a clear next step for the customer.
>
> Store policies: [INSERT POLICY DOCUMENT]
> Customer message: [INSERT CUSTOMER MESSAGE]

### Task 2 — Two key features matching the Functional style

1. **Precise, ordered task steps** (identify intent → extract details → check for missing info → resolve using policy) — this is a fixed procedure the model must execute exactly, not an open-ended conversation.
2. **Hard constraints on output** (use only provided policy text, ask only for the missing piece, under 100 words) — functional prompts prioritize accuracy and predictability over creativity or exploration.

### Task 3 — Justification

Customer support for e-commerce deals with real transactions (orders, refunds, money) where **accuracy and consistency matter more than personality or ideation**. An *exploratory* style would risk the bot inventing options that don't exist; a purely *conversational* style could sound warm but skip critical steps like verifying an order number; a *structured* style is close but is more about formatting than about a strict step-by-step decision procedure. The *functional* style ensures the bot behaves like a reliable process — extract data, check policy, act — which is exactly what a support chatbot handling real orders needs.

## 🌟 Exercise 5: Prompt Refinement Challenge — Control the Style, Structure, and Length

**Scenario:** PulseOne Mini smartwatch — email campaign blurb. Constraints: friendly tone, bullet points, max 50 words, must mention battery life, fitness tracking, and Bluetooth compatibility.

### Task 1 — Initial prompt

> Act as a marketing copywriter for a wearable tech brand. Write a product blurb for our new smartwatch, the **PulseOne Mini**, for an email campaign.
>
> - Tone: friendly and upbeat.
> - Format: bullet points.
> - Length: maximum 50 words total.
> - Must mention all 3 of these features: battery life, fitness tracking, and Bluetooth compatibility.

### Task 2 — Evaluation (run against a typical model)

Example output from running the initial prompt:

> - Meet PulseOne Mini — small watch, big energy! ⚡
> - Up to 7 days of battery life, so you barely need to charge.
> - Track steps, heart rate & workouts effortlessly.
> - Seamless Bluetooth pairing with your phone, always in sync.

**Checks:**
- Word count: this draft runs ~42 words — under 50, OK, but on a re-run the model sometimes drifts to 60–70 words if it adds extra flourishes (e.g., a closing tagline), so the constraint isn't always reliably obeyed.
- All 3 features mentioned: Yes (battery life, fitness tracking, Bluetooth) ✅.
- Tone: Friendly and engaging ✅, though occasionally the model adds a 4th or 5th bullet (e.g., a sign-off line) rather than sticking to exactly 3 feature bullets, which weakens length control.

### Task 3 — Revised prompt with stronger constraint language

> Act as a marketing copywriter for a wearable tech brand. Write a product blurb for our new smartwatch, the **PulseOne Mini**, for an email campaign.
>
> - Tone: friendly and upbeat.
> - **List exactly 3 bullet points — one for battery life, one for fitness tracking, one for Bluetooth compatibility. Do not add a 4th bullet, intro line, or closing tagline.**
> - **Do not exceed 50 words in total. Count your words before responding and cut anything over the limit.**
> - Each bullet must be a single short sentence.

## 🌟 Exercise 6: Hallucination Spotting and Mitigation

**Scenario:** Summarizing a peer-reviewed article on climate change and marine biodiversity; the model inserted an unsupported claim: *"Over 50% of marine species are projected to go extinct by 2050."*

### Task 1 — Prompt instructing the model to avoid hallucinations

> Summarize the following peer-reviewed article on climate change and marine biodiversity. **Use only information that is explicitly stated in the article text below. Do not add outside facts, statistics, or claims from general knowledge, even if they seem related or plausible.**
>
> Article: [PASTE ARTICLE TEXT]

### Task 2 — Rewritten with verification language

> Summarize the following peer-reviewed article on climate change and marine biodiversity.
>
> - **Only include claims explicitly stated in the article.** Do not infer, extrapolate, or supplement with outside knowledge.
> - **If a specific number, statistic, or projection is not explicitly given in the article, do not state one.** Instead, describe the finding qualitatively (e.g., "the article reports a decline in species diversity" rather than inventing a percentage).
> - **If you are uncertain whether a detail appears in the article, state that the information is not available** rather than guessing.
> - After the summary, list the direct quotes or sentences from the article that support each claim you made.
>
> Article: [PASTE ARTICLE TEXT]

### Task 3 — Two mitigation strategies used

1. **Explicit grounding/restriction instruction** — telling the model to use *only* the provided text and to avoid outside knowledge, which reduces the chance it pulls in plausible-sounding but unverified facts from its training data.
2. **Require source attribution (quote-back / cite evidence)** — asking the model to point to the specific sentence(s) in the article that support each claim. This makes fabrications easy to catch (if no supporting quote exists, the claim is likely hallucinated) and discourages the model from inventing unsupported statistics in the first place.

### Task 4 — Two professional domains where hallucinations are especially risky

1. **Healthcare** — A hallucinated drug dosage, contraindication, or diagnostic claim could directly harm a patient. Clinicians or patients relying on an AI summary of medical literature need every stated fact to be traceable to a real source, since incorrect medical information can lead to dangerous treatment decisions.
2. **Legal** — A hallucinated case citation, statute, or contract clause (a well-documented failure mode, e.g., lawyers citing fake case law) can mislead legal arguments, get filings sanctioned by courts, or cause a client to rely on non-existent legal protections. Legal work requires verifiable, real citations because outcomes have binding real-world consequences.
