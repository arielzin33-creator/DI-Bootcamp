# Daily Challenge: Precision Prompting for Control & Output Quality
Author: arielzin33@gmail.com

Scenario: turning a dense internal policy paragraph on remote access security into clear, employee-facing microlearning content, using a full design → generate → evaluate → refine prompt development cycle.

## Input Text (Original Policy Excerpt)

> Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications.

---
## Step 1: Craft a Prompt That Controls Output Tone, Format, and Length


### Prompt

> Act as an internal communications specialist writing a microlearning snippet for a company newsletter. Rewrite the policy text below in a **friendly, clear, and approachable tone** for a general (non-technical) employee audience.
>
> - **Paraphrase only** — do not copy any phrases directly from the original text.
> - Format the output as **bullet points** (2–3 bullets).
> - Keep the **total output under 75 words**.
> - Preserve every security requirement in the original — do not drop or soften any rule.
>
> Text: "Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."

### Simulated model output

> **Working remotely? Keep it secure:**
> - Always connect through the company's approved VPN when accessing internal systems from outside the office.
> - Never use public Wi-Fi or personal devices without proper security software to reach sensitive company data or communications.

*(34 words — well under the 75-word limit.)*

---
## Step 2: Evaluate the Output

| Criteria | Guiding Question | Evaluation |
|---|---|---|
| **Relevance** | Does the output stick to the source message? | Yes — it covers only VPN use and the ban on unsecured/personal devices, matching the source. |
| **Clarity** | Easier to understand for non-technical staff? | Yes — "public Wi-Fi," "security software" are simpler than "unsecured connections," "endpoint protection." |
| **Structure** | Formatted into clear bullet points? | Yes — a header line plus 2 bullets. |
| **Tone** | Appropriate for internal employee communication? | Yes — friendly, direct, "Keep it secure" reads as approachable newsletter copy. |
| **Length** | Under 75 words? | Yes — 34 words. |
| **Factual Accuracy** | Did the model add or remove critical security details? | ⚠️ Minor issue: the rewrite says **"public Wi-Fi"**, which is not stated in the original ("unsecured connections" is broader than just public Wi-Fi) — this is a small hallucinated specificity that narrows/alters the original meaning and should be flagged for correction. |

## Step 3: Detect and Mitigate Hallucinations

**Detected issue:** the model output introduced "public Wi-Fi" as an example, which is not in the source text ("unsecured connections" is the actual, broader term) — a small but real hallucination that could mislead an employee into thinking home Wi-Fi is automatically fine as long as it's not "public."

### Revised prompt (hallucination-mitigated)

> Act as an internal communications specialist writing a microlearning snippet for a company newsletter. Rewrite the policy text below in a **friendly, clear, and approachable tone** for a general (non-technical) employee audience.
>
> - **Paraphrase only** — do not copy phrases directly from the original text.
> - **Only paraphrase the content in the input text. Do not add new recommendations, examples, technologies, or specific scenarios (e.g., do not mention "public Wi-Fi," specific software names, or any detail not explicitly present in the input).**
> - Format the output as bullet points (2–3 bullets).
> - Keep the total output under 75 words.
> - Preserve every security requirement in the original — do not drop or soften any rule.
>
> Text: "Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."

### Simulated corrected output

> **Working remotely? Keep it secure:**
> - Always connect to internal systems through the company's approved VPN when working remotely.
> - Never use unsecured connections or personal devices without proper endpoint protection to access company data or sensitive communications.

*(35 words — accurate to source, no invented examples.)*

## Step 4: Paraphrasing Deep Dive (Junior Intern Audience)


### Prompt

> Act as a friendly onboarding buddy explaining a security rule to a **junior intern** who is new to corporate IT policies. Paraphrase the policy text below using **plain language and short phrases** — imagine explaining it to someone in their first week.
>
> - Use **no more than 4 bullet points**.
> - **Avoid all corporate or legal jargon** (e.g., don't use words like "proprietary," "endpoint protection," "under no circumstances" — say it simply instead).
> - Keep the tone **supportive and informative**, not scolding.
> - Do not add any information not present in the original text.
>
> Text: "Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."

### Simulated output

> **Staying safe when working remotely:**
> - Always use the company VPN to connect from outside the office.
> - Skip public or unprotected networks when reaching company systems.
> - Only use devices that have proper security tools installed.
> - This helps keep company data and messages safe.

*(4 bullets, plain wording, no jargon, supportive tone.)*

## Step 5: Quote Extraction Variant


### Prompt

> Read the policy text below and **extract the single direct quote (verbatim, unmodified) that best captures the core security requirement**. Do not paraphrase or alter the wording — copy it exactly as written, and present it in quotation marks. Do not add commentary beyond one short sentence explaining why this is the core rule.
>
> Text: "Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."

### Simulated output

> **Core quote:** "Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."
>
> This is the core rule because it states the hard prohibition (not just a recommendation) that defines the policy's actual enforcement boundary.

### When is quoting more appropriate than paraphrasing?

Quoting is better in **formal or compliance-sensitive internal communications** — e.g., official policy memos, legal/HR notices, audit documentation, or a compliance training module where employees may later be held to the *exact* wording of a rule. Verbatim quotes remove any risk that a paraphrase subtly shifts the legal or procedural meaning, and they let readers point back to the authoritative source text if there's ever a dispute about what was actually required.

### When might quoting pose a risk?

Quoting risks **losing accessibility and engagement** in casual or high-reach communications like a newsletter, chat announcement, or onboarding message — dense legal/technical phrasing (e.g., "endpoint protection," "under no circumstances") can confuse or disengage non-technical staff, defeating the purpose of a microlearning snippet. Quoting also risks **staying outdated**: if a quoted clause is later revised in the source policy, every place that quoted it verbatim must be manually updated, whereas a paraphrase can be more easily kept in sync with the policy's intent even as exact wording evolves.
